```python
# ---------------------------------------------------
# NETFLIX DATA PROJECT — BEGINNER FRIENDLY VERSION
# ---------------------------------------------------

# STEP 1: Import pandas library
import pandas as pd


# STEP 2: Read the Netflix CSV file
netflix_data = pd.read_csv(
    r'D:\MiniForge\Ex_Files_Intro_to_Data_Science\Ex_Files_Intro_to_Data_Science\Exercise Files\netflix_data.csv'
)


# STEP 3: Display the first 5 rows
netflix_data.head()
```

```python
# ---------------------------------------------------
# CHECK THE SHAPE OF THE DATASET
# ---------------------------------------------------

# Display the number of rows and columns
netflix_data.shape
```

```python
# ---------------------------------------------------
# DISPLAY COLUMN NAMES
# ---------------------------------------------------

# Display all column names
netflix_data.columns
```

```python
# ---------------------------------------------------
# CHECK DATA TYPES
# ---------------------------------------------------

# Display data types for each column
netflix_data.dtypes
```

```python
# ---------------------------------------------------
# CHECK FOR MISSING VALUES
# ---------------------------------------------------

# Count missing values in each column
netflix_data.isna().sum()
```

```python
# ---------------------------------------------------
# COUNT TOTAL ROWS WITH MISSING VALUES
# ---------------------------------------------------

# Count rows containing at least one missing value
netflix_data.isna().any(axis=1).sum()
```

```python
# ---------------------------------------------------
# REMOVE ROWS WITH MISSING VALUES
# ---------------------------------------------------

# Create a cleaned version of the dataset
clean_netflix = netflix_data.dropna()
```

```python
# ---------------------------------------------------
# DISPLAY SHAPE AFTER CLEANING
# ---------------------------------------------------

# Display the new shape after removing missing values
clean_netflix.shape
```

```python
# ---------------------------------------------------
# CHECK FOR DUPLICATE ROWS
# ---------------------------------------------------

# Count duplicate rows
netflix_data.duplicated().sum()
```

```python
# ---------------------------------------------------
# REMOVE DUPLICATE ROWS
# ---------------------------------------------------

# Remove duplicate rows
netflix_data = netflix_data.drop_duplicates()
```

```python
# ---------------------------------------------------
# CHECK DATA TYPE OF date_added COLUMN
# ---------------------------------------------------

# Display the data type of date_added
netflix_data['date_added'].dtype
```

```python
# ---------------------------------------------------
# CONVERT date_added TO DATETIME FORMAT
# ---------------------------------------------------

# Convert date_added column from string to datetime
netflix_data['date_added'] = pd.to_datetime(netflix_data['date_added'])
```

```python
# ---------------------------------------------------
# VERIFY NEW DATA TYPE
# ---------------------------------------------------

# Check the updated data type
netflix_data['date_added'].dtype
```

```python
# ---------------------------------------------------
# DISPLAY THE EARLIEST RELEASE YEAR
# ---------------------------------------------------

# Display the minimum release year
netflix_data['release_year'].min()
```

```python
# ---------------------------------------------------
# DISPLAY THE LATEST RELEASE YEAR
# ---------------------------------------------------

# Display the maximum release year
netflix_data['release_year'].max()
```

```python
# ---------------------------------------------------
# SORT MOVIES BY RELEASE YEAR
# ---------------------------------------------------

# Sort release years in ascending order
netflix_data.sort_values('release_year')
```

```python
# ---------------------------------------------------
# FILTER ONLY MOVIES
# ---------------------------------------------------

# Create a DataFrame containing only movies
movies = netflix_data.loc[netflix_data['type'] == 'Movie', :]
```

```python
# ---------------------------------------------------
# DISPLAY MOVIES DATAFRAME
# ---------------------------------------------------

movies
```

```python
# ---------------------------------------------------
# FILTER ONLY TV SHOWS
# ---------------------------------------------------

# Create a DataFrame containing only TV shows
tv_shows = netflix_data.loc[netflix_data['type'] == 'TV Show', :]
```

```python
# ---------------------------------------------------
# DISPLAY TV SHOWS DATAFRAME
# ---------------------------------------------------

tv_shows
```

```python
# ---------------------------------------------------
# FIND TITLES RELEASED IN 2020
# ---------------------------------------------------

# Filter titles released in 2020
titles_2020 = netflix_data.loc[netflix_data['release_year'] == 2020, :]
```

```python
# ---------------------------------------------------
# DISPLAY TITLES RELEASED IN 2020
# ---------------------------------------------------

titles_2020
```

```python
# ---------------------------------------------------
# SORT TITLES BY RELEASE YEAR DESCENDING
# ---------------------------------------------------

# Sort titles from newest to oldest
sorted_titles = netflix_data.sort_values('release_year', ascending=False)
```

```python
# ---------------------------------------------------
# DISPLAY SORTED TITLES
# ---------------------------------------------------

sorted_titles
```

```python
# ---------------------------------------------------
# GROUP DATA BY TYPE
# ---------------------------------------------------

# Count the number of Movies and TV Shows
netflix_data.groupby('type').count()
```
