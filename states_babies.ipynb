```python
# Create a new DataFrame containing only California baby name records
ca_babies = state_babies.loc[state_babies['State'] == 'CA', :]
```

```python
ca_babies
```

```python
# Sort California baby names from highest Count to lowest Count
ca_babies_sorted = ca_babies.sort_values('Count', ascending=False)
```

```python
# Define a function that returns the first name in each grouped list
# Since the data is already sorted by Count, the first name is the most popular
def popular(s):
    return s.iloc[0]
```

```python
# Group California baby names by Year and Gender
# Then apply the popular function to find the most popular name in each group
ca_grouped = ca_babies_sorted.groupby(['Year', 'Gender']).agg(popular)
```
