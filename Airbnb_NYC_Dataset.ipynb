```python
# --------------------------------------------------
# IMPORT REQUIRED LIBRARIES
# --------------------------------------------------

# Import pandas for working with data
import pandas as pd

# Import matplotlib for charts and graphs
import matplotlib.pyplot as plt

# Import seaborn for data visualization
import seaborn as sn
```

```python
# --------------------------------------------------
# READ THE AIRBNB NYC DATASET
# --------------------------------------------------

# Load the Airbnb NYC 2019 dataset into a DataFrame
listings = pd.read_csv(
    r'D:\MiniForge\Ex_Files_Intro_to_Data_Science\Ex_Files_Intro_to_Data_Science\Exercise Files\Airbnb_NYC_2019.csv'
)
```

```python
# --------------------------------------------------
# DISPLAY THE FIRST 5 ROWS
# --------------------------------------------------

# Display the first 5 rows of the dataset
listings.head()
```

```python
# --------------------------------------------------
# DISPLAY THE ENTIRE DATAFRAME
# --------------------------------------------------

# Display the Airbnb listings dataset
listings
```

```python
# --------------------------------------------------
# CREATE A BAR CHART SHOWING THE COUNT OF AIRBNB
# LISTINGS IN EACH NEIGHBORHOOD GROUP
# --------------------------------------------------

# Create a count plot for neighborhood groups
sn.countplot(x='neighbourhood_group', data=listings)

# Display the chart
plt.show()
```

```python
# --------------------------------------------------
# CREATE A BAR CHART SHOWING THE AVERAGE PRICE
# OF AIRBNB LISTINGS IN EACH NEIGHBORHOOD GROUP
# --------------------------------------------------

# Create a bar plot for average prices
sn.barplot(
    x='neighbourhood_group',
    y='price',
    data=listings
)

# Display the chart
plt.show()
```

```python
# --------------------------------------------------
# CREATE THE SAME BAR CHART WITHOUT
# CONFIDENCE INTERVAL LINES
# --------------------------------------------------

# Create a bar plot without confidence intervals
sn.barplot(
    x='neighbourhood_group',
    y='price',
    data=listings,
    ci=False
)

# Display the chart
plt.show()
```

```python
# --------------------------------------------------
# SORT LISTINGS BY PRICE FROM HIGHEST TO LOWEST
# --------------------------------------------------

# Sort the Airbnb listings by price in descending order
sorted_prices = listings.sort_values(
    'price',
    ascending=False
)

# Display the sorted data
sorted_prices
```

```python
# --------------------------------------------------
# DISPLAY ONLY THE TOP 10 MOST EXPENSIVE LISTINGS
# --------------------------------------------------

# Display the first 10 rows from the sorted dataset
sorted_prices.iloc[0:10]
```
