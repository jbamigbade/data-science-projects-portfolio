```python
# --------------------------------------------------
# IMPORT REQUIRED LIBRARIES
# --------------------------------------------------

# Import pandas for working with data
import pandas as pd

# Import matplotlib for charts and graphs
import matplotlib.pyplot as plt

# Import numpy for numerical operations
import numpy as np
```

```python
# --------------------------------------------------
# READ THE AIRBNB NYC DATASET
# --------------------------------------------------

# Load the Airbnb NYC 2019 dataset into a DataFrame
listings = pd.read_csv(
    r'D:\MiniForge\Ex_Files_Intro_to_Data_Science\Ex_Files_Intro_to_Data_Science\Exercise Files\Airbnb_NYC_2019.csv'
)
```

```python
# --------------------------------------------------
# DISPLAY THE FIRST 5 ROWS
# --------------------------------------------------

# Display the first 5 rows of the dataset
listings.head()
```

```python
# --------------------------------------------------
# DISPLAY THE ENTIRE DATAFRAME
# --------------------------------------------------

# Display the Airbnb listings dataset
listings
```

```python
# --------------------------------------------------
# CREATE A HISTOGRAM OF AIRBNB PRICES
# --------------------------------------------------

# Create a histogram showing the distribution of Airbnb prices
plt.hist(listings['price'])

# Label the x-axis
plt.xlabel('Price (in US dollars)')

# Display the histogram
plt.show()
```

```python
# --------------------------------------------------
# CREATE A HISTOGRAM WITH CUSTOM BINS
# --------------------------------------------------

# Create a histogram with custom bin sizes
plt.hist(
    listings['price'],
    bins=np.arange(0, 1100, 40)
)

# Label the x-axis
plt.xlabel('Price (in US dollars)')

# Display the histogram
plt.show()
```

```python
# --------------------------------------------------
# CREATE A SCATTER PLOT OF PRICE VS NUMBER OF REVIEWS
# --------------------------------------------------

# Create a scatter plot comparing price and number of reviews
plt.scatter(
    x=listings['price'],
    y=listings['number_of_reviews']
)

# Label the x-axis
plt.xlabel('Price')

# Label the y-axis
plt.ylabel('Number of Reviews')

# Display the scatter plot
plt.show()
```

```python
# --------------------------------------------------
# CREATE A SCATTER PLOT WITH X-AXIS LIMIT
# --------------------------------------------------

# Create a scatter plot with a restricted x-axis
plt.scatter(
    x=listings['price'],
    y=listings['number_of_reviews']
)

# Label the x-axis
plt.xlabel('Price')

# Label the y-axis
plt.ylabel('Number of Reviews')

# Restrict the x-axis from 0 to 1100
plt.xlim(0, 1100)

# Display the scatter plot
plt.show()
```

```python
# --------------------------------------------------
# CREATE A SCATTER PLOT WITH SMALLER POINTS
# --------------------------------------------------

# Create a scatter plot with smaller point sizes
plt.scatter(
    x=listings['price'],
    y=listings['number_of_reviews'],
    s=5
)

# Label the x-axis
plt.xlabel('Price')

# Label the y-axis
plt.ylabel('Number of Reviews')

# Restrict the x-axis from 0 to 1100
plt.xlim(0, 1100)

# Display the scatter plot
plt.show()
```

```python
# --------------------------------------------------
# SORT AIRBNB LISTINGS BY PRICE
# --------------------------------------------------

# Sort the listings by price from highest to lowest
sorted_prices = listings.sort_values(
    'price',
    ascending=False
)

# Display the sorted listings
sorted_prices
```
